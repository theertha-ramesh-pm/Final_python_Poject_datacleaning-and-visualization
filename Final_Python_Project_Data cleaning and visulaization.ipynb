<h3>Importing Required Libraries</h3>

Necessary Python libraries such as Pandas, NumPy, Matplotlib, and Seaborn are used for data analysis and visualization.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("student-por.csv", sep=",")

df.head()
```


<h2>Section I: Baseline Student Performance Analysis</h2>

This section establishes a foundational understanding of student academic performance. It presents visualizations of final grades and grade distributions to provide an overview of students' academic outcomes. These visualizations serve as a reference point for subsequent analyses of study habits, attendance, and other factors associated with academic performance.



### (i) Distribution of Final Grades

- **Visualization Type:** Histogram (Seaborn)
- **Purpose:** To analyze the distribution of students' final grades (G3).
- **Insight:** The chart helps identify the grade ranges in which students are concentrated and provides an overview of academic performance.


```python

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="G3",
    bins=21,
    kde=True
)

plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```


### (ii) Comparison of G1, G2 and G3 Grades

- **Visualization Type:** Box Plot (Seaborn)
- **Purpose:** To compare the distributions of first-period, second-period, and final grades.
- **Insight:** The chart helps compare median grades, variation, and potential outliers across the three grade variables.


```python

grade_data = df[["G1", "G2", "G3"]].melt(
    var_name="Grade Type",
    value_name="Grade"
)

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=grade_data,
    x="Grade Type",
    y="Grade"
)

plt.title("Comparison of G1, G2 and G3 Grades")
plt.xlabel("Grade Type")
plt.ylabel("Grade (0–20)")
plt.tight_layout()
plt.show()
```


<h2>Section II: Study Habits and Academic Performance</h2>

This section explores the relationship between students' study habits, attendance, previous academic failures, and final grades. The visualizations help identify patterns in academic performance across different study-time categories and previous-failure groups.



<h3>(iii) Average Final Grade by Study Time</h3>

- <b>Visualization Type:</b> Bar Chart (Seaborn)
- <b>Purpose:</b> To compare average final grades across study-time categories.
- <b>Insight:</b> The chart helps identify differences in average academic performance across study-time categories.


```python

study_performance = (
    df.groupby("studytime", as_index=False)["G3"]
      .mean()
)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=study_performance,
    x="studytime",
    y="G3"
)

plt.title("Average Final Grade by Study Time")
plt.xlabel("Study Time Category")
plt.ylabel("Average Final Grade")
plt.tight_layout()
plt.show()
```


<h3>(iv) Relationship Between Absences and Final Grade</h3>

- <b>Visualization Type:</b> Scatter Plot (Seaborn)
- <b>Purpose:</b> To examine the relationship between student absences and final grades.
- <b>Insight:</b> The chart helps identify patterns between attendance and academic performance.


```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="absences",
    y="G3",
    alpha=0.6
)

plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade (G3)")
plt.tight_layout()
plt.show()
```


<h3>(v) Previous Failures vs Final Grade</h3>

- <b>Visualization Type:</b> Bar Chart (Seaborn)
- <b>Purpose:</b> To compare average final grades across previous-failure categories.
- <b>Insight:</b> The chart highlights differences in average final grades among these categories.


```python

failure_performance = (
    df.groupby("failures", as_index=False)["G3"]
      .mean()
)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=failure_performance,
    x="failures",
    y="G3"
)

plt.title("Average Final Grade by Previous Failures")
plt.xlabel("Number of Previous Failures")
plt.ylabel("Average Final Grade")
plt.tight_layout()
plt.show()
```


<h3>(vi) Study Time and Previous Failures vs Final Grade</h3>

- <b>Visualization Type:</b> Heatmap (Seaborn)
- <b>Purpose:</b> To compare average final grades across combinations of study time and previous failures.
- <b>Insight:</b> The heatmap highlights variations in average final grades across the combinations.


```python

study_failure = df.pivot_table(
    index="studytime",
    columns="failures",
    values="G3",
    aggfunc="mean"
)

plt.figure(figsize=(8, 5))

sns.heatmap(
    study_failure,
    annot=True,
    fmt=".2f",
    cmap="YlGnBu"
)

plt.title("Study Time and Previous Failures vs Final Grade")
plt.xlabel("Number of Previous Failures")
plt.ylabel("Study Time Category")
plt.tight_layout()
plt.show()
```


<h2>Section III: Student Characteristics Analysis</h2>

This section examines student performance across different demographic and school-related categories. The visualizations compare grade distributions by gender and average final grades across schools to identify differences in academic outcomes.



<h3>(vii) Gender-wise Final Grade Comparison</h3>

- <b>Visualization Type:</b> Box Plot (Seaborn)
- <b>Purpose:</b> To compare final grade distributions across gender categories.
- <b>Insight:</b> The chart highlights differences in grade distributions between the categories.


```python

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="sex",
    y="G3"
)

plt.title("Gender-wise Final Grade Comparison")
plt.xlabel("Gender")
plt.ylabel("Final Grade (G3)")
plt.tight_layout()
plt.show()
```


<h3>(viii) School-wise Average Final Grade</h3>

- <b>Visualization Type:</b> Bar Chart (Seaborn)
- <b>Purpose:</b> To compare average final grades across school categories.
- <b>Insight:</b> The chart helps identify differences in average academic performance between schools.


```python

school_performance = (
    df.groupby("school", as_index=False)["G3"]
      .mean()
)

plt.figure(figsize=(7, 5))

sns.barplot(
    data=school_performance,
    x="school",
    y="G3"
)

plt.title("School-wise Average Final Grade")
plt.xlabel("School")
plt.ylabel("Average Final Grade")
plt.tight_layout()
plt.show()
```


<h2>Section IV: Grade Relationships and Correlation Analysis</h2>

This section investigates relationships among students' academic grades and selected numerical variables. Scatter plots and correlation heatmaps are used to examine grade patterns and identify positive or negative linear relationships among the variables.



<h3>(ix) Relationship Between G1 and G3 Grades</h3>

- <b>Visualization Type:</b> Scatter Plot with Regression Line (Seaborn)
- <b>Purpose:</b> To examine the relationship between first-period grades and final grades.
- <b>Insight:</b> The chart helps identify whether higher first-period grades tend to be associated with higher final grades.


```python

plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="G1",
    y="G3",
    scatter_kws={"alpha": 0.5},
    line_kws={"color": "red"}
)

plt.title("Relationship Between G1 and G3 Grades")
plt.xlabel("First-period Grade (G1)")
plt.ylabel("Final Grade (G3)")
plt.tight_layout()
plt.show()
```


<h3>(x) Correlation Heatmap of Student Performance</h3>

- <b>Visualization Type:</b> Correlation Heatmap (Seaborn)
- <b>Purpose:</b> To examine correlations among selected numerical variables related to student performance.
- <b>Insight:</b> The heatmap helps identify the strength and direction of linear relationships among the variables.


```python

corr_columns = [
    "age",
    "studytime",
    "failures",
    "absences",
    "G1",
    "G2",
    "G3"
]

correlation_matrix = df[corr_columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap of Student Performance")
plt.tight_layout()
plt.show()
```
